```python
import threading
import time
##1st thread
def print_numbers():
  for i in range (5):
    time.sleep(2)
    print(f"Numbers:{i}")
##2nd thread
def print_letter():
  for letter in "abcde":
    time.sleep(2)
    print(f"letter:{letter}")
##create 2 threads
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_letter)

t=time.time()

##start the thread
t1.start()
t2.start()


##wait for threads to complete
t1.join()
t2.join()
fin_time=time.time()-t
print(fin_time)
```

```python
##Multiprocessing

import multiprocessing
import time

def square_no():
  for i in range(5):
    time.sleep(1)
    print(f"square: {i*i}")

def cube_no():
  for i in range(5):
    time.sleep(1.5)
    print(f"cube: {i*i*i}")

if __name__=="__main__":
  ##create 2 proc

  p1=multiprocessing.Process(target=square_no)
  p2=multiprocessing.Process(target=cube_no)
  t=time.time()
  ##start the proc
  p1.start()
  p2.start()

  ##wait
  p1.join()
  p2.join()

  fin_time=time.time()-t
  print(fin_time)
```

With ThreadPoolExecutor, multiple tasks can run simultaneously,many tasks can run together


executor: manager that create,assign,collect res etc

map: apply same func to many inputs
map(function,iterable)

```python
## using threadpool executor

from concurrent.futures import ThreadPoolExecutor
import time

def print_no(numbers):
  time.sleep(1)
  return f"numbers : {numbers}"

numbers=[1,2,3,4,5,6,7,8,4,5,3,5,6]

with ThreadPoolExecutor(max_workers=3) as executor:
  results=executor.map(print_no,numbers)

for result in results:
  print(result)
```

```python
##multiprocessing with processpoolexecutor


from concurrent.futures import ProcessPoolExecutor
import time

def square_no(numbers):
  time.sleep(5)
  return f"numbers : {numbers * numbers}"

numbers=[1,2,3,4,5]

with ProcessPoolExecutor(max_workers=3) as executor:
  results=executor.map(square_no,numbers)

t = time.time()
print(t)

for result in results:
  print(result)
```
